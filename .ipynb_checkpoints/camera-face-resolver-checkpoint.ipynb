```python
import cv2
```

```python
url = "http://192.168.1.102:8081/video"
cap = cv2.VideoCapture(url)

if not cap.isOpened():
    print("Cannot connect to stream")
```

```python
while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Display the frame in a separate window
    cv2.imshow('Rubiks Stream', frame)

    # Press 'q' to quit the window
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
